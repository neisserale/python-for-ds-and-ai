## Ejercicio 5: Simular el entrenamiento de un modelo

Un modelo se entrena durante varias iteraciones llamadas **epochs**.

Crea una función que reciba:

- `loss` inicial
- `loss_reduction_rate`
- Número máximo de `epochs`

El `loss` representa el error del modelo. Para simplificar el entrenamiento, asumiremos que el `loss` se reduce en un `20%` después de cada epoch.

El nuevo `loss` se calcula como:

```python
new_loss = loss - loss x loss_reduction_rate
new_loss = loss × (1 - loss_reduction_rate)
```

El entrenamiento debe ejecutarse durante un máximo de `epochs` y detenerse antes si el `loss` es menor o igual a `0.1`.

**Ejemplo:**

```python
initial_loss = 1.0
decay_rate = 0.2
total_epochs = 20

train_model(loss=initial_loss, loss_reduction_rate=decay_rate, epochs=total_epochs)
```

**Salida esperada:**

```text
Epoch 1: Loss = 0.800
Epoch 2: Loss = 0.640
Epoch 3: Loss = 0.512
...
Epoch 11: Loss = 0.086

Training completed
```

> **Pista:** Usa `for`, `range()`, `if` y `break`.
>
> **Nota:** Esta es una simulación simplificada. En un entrenamiento real, el `loss` se calcula a partir de las predicciones del modelo.
