# Comparison Operators

* **Equality:** `a == b`
* **Inequality:** `a != b`
* **Less than:** `a < b`
* **Less than or equal to:** `a <= b`
* **Greater than:** `a > b`
* **Greater than or equal to:** `a >= b`

```python
a = 5
b = 10

print(a == b)
print(a != b)
print(a < b)
print(a <= b)
print(a > b)
print(a >= b)
```

```python
## negating a comparison result
a < b, not a < b
```

## If - Else

```python
score = 10.45

if score >= 10.5:
    print('Approved')

else:
    print("Desapproved")
```

```python
age = 40

if age >= 18 and age < 65:
    print('Working age')

else:
    print('Other')
```

## Elif

```python
score = 15

if score >= 17:
    print("Excellent")

elif score >= 14:
    print("Good")
    
else:
    print("Needs improvement")
```

```python
x = 10

if x > 5:
    print("Greater than 5")

elif x > 8:
    print("Greater than 8")

else:
    print("Greater than 12")
```

```python
x = 10

if x > 5:
    print("Greater than 5")

if x > 8:
    print("Greater than 8")

if x > 12:
    print("Greater than 12")
```

## Nested

```python
score = 19

if score >= 10.5:
    print('Approved')

    if score >= 17:
        print("Excellent")

    elif score >= 14:
        print("Good")
        
    else:
        print("Needs improvement")

else:
    print("Desapproved")
```

## One Line If

```python
result = 'Approved' if score >= 10.5 else 'Disapproved'
result
```

# Loops

## For

```python
range(5)
```

```python
for number in range(10):
    print(number)
```

```python
for i, number in enumerate(range(10, 20)):
    print(f'index {i}: {number}')
```

```python
for number in range(10):
    power = number ** 2
    print(f'Square of {i} -> {power}')
```

```python
for number in range(10):
    if number % 2 == 0:
        print(number)
```

```python
for number in range(10):   
    if number == 5:
        break

    print(number)
```

```python
for number in range(10):
    if number == 5:
        continue

    print(number)
```

```python
for i in range(2):
    for j in range(5):
        print(f'Position: {i}, {j}')
```

## While

```python
i = 1
while i < 10:
    print(i)
    i += 1
```

```python
i = 1
while i < 10:
    print(i)

    if i == 5:
        break

    i += 1
```

```python
i = 1
while True:
    print(i)

    if i == 5:
        break

    i += 1
```
