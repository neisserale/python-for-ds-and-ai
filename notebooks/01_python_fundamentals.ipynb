## If - Else

Adds a fallback branch that runs when the condition is False

> ```python
> if condition:
>     ## runs if True
> else:
>     ## runs if False

````python
score = 10.45

if score >= 10.5:
    print('Approved')

else:
    print("Desapproved")
````

````python
age = 40

if age >= 18 and age < 65:
    print('Working age')

else:
    print('Other')
````

## Elif

Checks multiple conditions in sequence. Only the first condition that evaluates to True is executed

> ```python
> if condition_1:
>     ## runs if condition_1 is True
> elif condition_2:
>     ## runs if condition_1 is False and condition_2 is True
> else:
>     ## runs if none of the above are True


````python
score = 15

if score >= 17:
    print("Excellent")

elif score >= 14:
    print("Good")
    
else:
    print("Needs improvement")
````

````python
x = 10

if x > 5:
    print("Greater than 5")

elif x > 8:
    print("Greater than 8")

else:
    print("Greater than 12")
````

````python
x = 10

if x > 5:
    print("Greater than 5")

if x > 8:
    print("Greater than 8")

if x > 12:
    print("Greater than 12")
````

## Nested

An if inside another if, for conditions that depend on each other

> ```python
> if condition_1:
>     if condition_2:
>         ## runs only if both conditions are True


````python
score = 19

if score >= 10.5:
    print('Approved')

    if score >= 17:
        print("Excellent")

    elif score >= 14:
        print("Good")
        
    else:
        print("Needs improvement")

else:
    print("Desapproved")
````

# Loops

## For

Iterates over a sequence (like range()) a fixed number of times, or until break

> ```python
> for item in sequence:
>     ## do something to each item


````python
range(5)
````

````python
for number in range(10):
    print(number)
````

````python
for i, number in enumerate(range(10, 20)):
    print(f'index {i}: {number}')
````

````python
for number in range(10):
    power = number ** 2
    print(f'Square of {number} -> {power}')
````

````python
for number in range(10):
    if number % 2 == 0:
        print(number)
````

````python
for number in range(10):   
    if number == 5:
        break

    print(number)
````

````python
for number in range(10):
    if number == 5:
        continue

    print(number)
````

````python
for i in range(2):
    for j in range(5):
        print(f'Position: {i}, {j}')
````

## While

Repeats a block as long as a condition stays True

> ```python
> while condition:
>     ## do something repeatedly


````python
i = 1
while i < 10:
    print(i)
    i += 1
````

````python
i = 1
while i < 10:
    print(i)

    if i == 5:
        break

    i += 1
````

````python
print('You need to enter "exit" to close the system.')

while True:
    command = input('Enter command: ')

    if command == 'exit':
        print('Closing the system...')
        break

    print(f'You entered: {command}')
````

# Try / Except

`try / except` is used to handle errors without stopping the program.

> ```python
> try:
>     ## try to execute this code
>
> except:
>     ## if an error occurs, execute this code

````python
## re-raising an exception
try:
    value = float('hello')

except:
    raise
````

````python
## generic exception
try:
    value = float('hello')

except:
    print('Something went wrong')
````

````python
## specific exception
try:
    value = float('hello')

except ValueError as error:
    print(f'Invalid number: {error}')
````

# Functions

`functions` group reusable blocks of code that can be called with different inputs.

> ```python
> def function_name(parameters):
>     ## code to execute
>     return value

````python
def greet():
    print('Hello!')

## calling functions
greet()
````

````python
## 'name' is the parameter: the variable the function expects to receive
def greet_person(name):
    print(f'Hello, {name}')

## "Jhon" is the argument: the actual value passed when calling the function
greet_person('Jhon')
````

````python
## default parameters
def greet_person(name='guest'):
    print(f"Hello, {name}")

greet_person()      # uses the default value
greet_person('Ana') # overrides the default value
````

````python
## return
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
````

````python
## multiple return
def operations(a, b):
    total = a + b
    diff = a - b
    prod = a * b
    
    return total, diff, prod  # Python packs this as a tuple

total, diff, prod = operations(3, 5)
print(total, diff, prod)
````

## **kwargs

````python
## **kwargs groups extra keyword arguments into a dictionary
def show_info(**kwargs):
    print(kwargs)

show_info(name="Jhon", role="Developer", company="OpenAI")
````

````python
def train_model(model, **params):
    print(f'Model: {model}')
    print(f'Parameters: {params}')

train_model(
    'Random Forest',
    n_estimators=100,
    max_depth=5,
    random_state=42,
)
````

## Ejercicio 5: Simular el entrenamiento de un modelo

Un modelo se entrena durante varias iteraciones llamadas **epochs**.

Crea una función que reciba:

- `loss` inicial
- `loss_reduction_rate`
- Número máximo de `epochs`

El `loss` representa el error del modelo. Para simplificar el entrenamiento, asumiremos que el `loss` se reduce en un `20%` después de cada epoch.

El nuevo `loss` se calcula como:

```text
new_loss = loss - loss x loss_reduction_rate
new_loss = loss × (1 - loss_reduction_rate)
```

El entrenamiento debe ejecutarse durante un máximo de `epochs` y detenerse antes si el `loss` es menor o igual a `0.1`.

**Ejemplo:**

```python
initial_loss = 1.0
decay_rate = 0.2
total_epochs = 20

train_model(loss=initial_loss, loss_reduction_rate=decay_rate, epochs=total_epochs)
```

**Salida esperada:**

```text
Epoch 1: Loss = 0.800
Epoch 2: Loss = 0.640
Epoch 3: Loss = 0.512
...
Epoch 11: Loss = 0.086

Training completed
```

> **Pista:** Usa `for`, `range()`, `if` y `break`.
>
> **Nota:** Esta es una simulación simplificada. En un entrenamiento real, el `loss` se calcula a partir de las predicciones del modelo.
