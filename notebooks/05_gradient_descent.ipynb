# Gradient Descent

## Cost function

- A cost function $J(w)$ measures how wrong a model is for a given parameter $w$.
- We start with a simple one: $J(w) = (w - 3)^2$, whose minimum is at $w = 3$ with $J = 0$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
## define the cost function
def J(w):
    return (w - 3) ** 2
```

```python
## plot the cost function
ws = np.linspace(-2, 8, 100)

plt.plot(ws, J(ws))
plt.xlabel('w')
plt.ylabel('J(w)')
plt.title('J(w) = (w - 3)²')
plt.show()
```

## Minimizing with gradient descent

- The derivative $J'(w) = 2(w - 3)$ points uphill, so we step in the opposite direction.
- Update rule: $w \leftarrow w - \alpha \, J'(w)$, where $\alpha$ is the learning rate.

```python
## derivative of the cost function
def dJ(w):
    return 2 * (w - 3)
```

```python
## start far from the minimum and walk downhill
w = -2          ## initial point
alpha = 0.1     ## learning rate
history = [w]

for k in range(1, 31):
    w = w - alpha * dJ(w)
    history.append(w)
    print(k, round(w, 4), round(J(w), 4))
```

```python
## plot the gradient descent path on the cost curve
plt.plot(ws, J(ws), label='J(w)')
plt.plot(history, [J(w) for w in history], 'o-', color='red', label='gradient descent')
plt.xlabel('w')
plt.ylabel('J(w)')
plt.legend()
plt.show()
```

```python
x = 4        # punto inicial
alpha = 0.1   # learning rate

for k in range(1, 52):
    grad = 2 * x               # 1. derivada f'(x) = 2x
    x = x - alpha * grad       # 2. actualización
    print(k, round(x, 4), round(x**2, 4))
```

```python
X = [1, 2, 3]          # metros cuadrados
Y = [3, 5, 7]          # precio real
w, b = 0.5, 0
alpha = 0.1
n = len(X)

for k in range(1, 179):
    # 1. predicciones y errores (ŷ − y)
    errores = [w * x + b - y for x, y in zip(X, Y)]
    # 2. derivadas parciales del MSE
    dw = (2 / n) * sum(e * x for e, x in zip(errores, X))
    db = (2 / n) * sum(errores)
    # 3. actualización (las dos a la vez)
    w = w - alpha * dw
    b = b - alpha * db
    J = sum((w * x + b - y) ** 2 for x, y in zip(X, Y)) / n
    print(k, round(w, 4), round(b, 4), round(J, 5))
```
