# Lists

Lists are used to store multiple items in a single variable.

````python
numbers = [1, 2, 3, 4, 5]

print(numbers, type(numbers))
````

````python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
planets
````

````python
values = [1, 'python', 3.14, True, [10, 20, 30]]
values
````

## Indexing
You can access individual list elements with square brackets.

````python
planets[0]
````

````python
planets[2]
````

````python
planets[-1]
````

````python
'Earth' in planets
````

````python
'pluto' not in planets
````

## Slicing
Use `start:stop` to extract a sub-list (the `stop` index is excluded).

````python
planets[0:3]
````

````python
planets[:3]
````

````python
planets[3:]
````

````python
## the last 3 planets
planets[-3:]
````

````python
## all the planets except the first and last
planets[1:-1]
````

## Changing lists
Lists are "mutable", meaning they can be modified "in place".

````python
planets[3] = 'Red Planet'

planets
````

## List Methods

* **append():** Adds an element to the end of the list.
* **extend():** Appends elements from another list (or iterable) to the end of the list.
* **insert():** Inserts an element at a specific position.
* **remove():** Removes the first occurrence of a specific element.
* **pop():** Removes and returns the element at a specific position.
* **clear():** Removes all the elements from the list.

````python
languages = ['C#', 'Java', 'C++']
languages
````

````python
languages.append('Python')
languages
````

````python
languages.extend(['Go', 'R'])
languages
````

````python
languages.insert(0, 'Cobol')
languages
````

````python
languages.remove('R')
languages
````

````python
## pop() returns the removed element, not the list
languages.pop(0)
````

````python
languages.reverse()
languages
````

````python
languages.count('Python')
````

````python
languages.index('Python')
````

````python
my_languages = languages.copy()
my_languages
````

````python
languages.clear()
languages
````

## List Functions

* **len():** Returns the number of elements in the list.
* **sorted():** Returns a new sorted list, leaving the original unchanged.
* **sum():** Returns the sum of all the elements.
* **min() / max():** Return the smallest / largest element.

````python
ages = [10, 15, 35, 27, 50, 36, 68, 42, 20]
ages
````

````python
len(ages)
````

````python
sorted(ages)
````

````python
max(ages)
````

````python
min(ages)
````

````python
sum(ages)
````

## Iterating Lists
Loop through a list directly, with indices via `enumerate()`, or in parallel with another list via `zip()`.

````python
names = ['Robert', 'Taylor', 'Tom']

for n in names:
    print(n)
````

````python
for i, n in enumerate(names):
    print(i, n)
````

````python
ages = [30, 35, 20]

## zip() pairs up elements from both lists
for name, age in zip(names, ages):
    print(name, age)
````

# Dictionaries

Dictionaries are a built-in Python data structure for mapping keys to values.

````python
customers = {'Robert':30, 'Taylor':35, 'Tom':20}
customers, type(customers)
````

## Indexing
Access a value by its key using square brackets.

````python
customers['Robert']
````

````python
'Taylor' in customers
````

````python
'Mirra' in customers
````

## Changing dictionaries
Dictionaries are also mutable: you can add or update key-value pairs directly.

````python
customers['Robert'] = 40
customers
````

````python
customers['Mirra'] = 22
customers
````

## Dictionary Methods

* **keys():** Returns a view of all the dictionary's keys.
* **values():** Returns a view of all the dictionary's values.
* **items():** Returns a view of the dictionary's key-value pairs.
* **get():** Returns the value for a key, or `None` if the key doesn't exist.
* **update():** Adds or updates key-value pairs from another dictionary.
* **pop():** Removes a key and returns its value.
* **clear():** Removes all the keys and values from the dictionary.

````python
customers.keys()
````

````python
customers.values()
````

````python
customers.items()
````

````python
customers.get('Robert')
````

````python
customers.update({'Frank': 34})
customers
````

````python
## pop() returns the removed value, not the dictionary
customers.pop('Frank')
````

````python
customers
````

````python
customers.clear()
customers
````

## Iterating Dictionaries
By default, iterating a dictionary walks over its keys.

````python
books = {
    'The Odyssey':'Homer',
    'Hamlet':'William Shakespeare',
    'The Divine Comedy':'Dante Alighieri',
    'Crime and Punishment':'Fyodor Dostoevsky',
    'Les Misérables':'Victor Hugo'
}

books
````

````python
for b in books:
    print(b)
````

````python
for b in books:
    print(books[b])
````

````python
for k, v in books.items():
    print(k, '-', v)
````

# Data Structure Comprehensions

Comprehensions build a new list, set, or dictionary in a single readable line, instead of writing a full `for` loop.

## List Comprehensions

> ```python
> [expression for item in iterable if condition]
> ```

````python
squares = []
for n in range(10):
    squares.append(n**2)
    
squares
````

````python
squares = [n**2 for n in range(10)]
squares
````

````python
squares = [n**2 for n in range(10) if n%2 == 0]
squares
````

## Dictionary Comprehensions

> ```python
> {key_expression: value_expression for item in iterable if condition}
> ```

````python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']

planet_to_initial = {planet: planet[0] for planet in planets}
planet_to_initial
````
