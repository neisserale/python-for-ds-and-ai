# Lists

Lists are used to store multiple items in a single variable.

```python
numbers = [1, 2, 3, 4, 5]

print(numbers, type(numbers))
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
planets
```

```python
values = [1, 'python', 3.14, True, [10, 20, 30]]
values
```

## Indexing
You can access individual list elements with square brackets.

```python
planets[0]
```

```python
planets[2]
```

```python
planets[-1]
```

## Slicing

```python
planets[0:3]
```

```python
planets[:3]
```

```python
planets[3:]
```

```python
# the last 3 planets
planets[-3:]
```

```python
# all the planets except the first and last
planets[1:-1]
```

## Changing lists
Lists are "mutable", meaning they can be modified "in place".

```python
planets[3] = 'Red Planet'

planets
```

## List Methods

* append(): Adds an element to the end of the list.
* extend(): Appends elements from another list (or iterable) to the end of the list.
* insert(): Inserts an element at a specific position.
* remove(): Removes the first occurrence of a specific element.
* pop(): Removes and returns the element at a specific position.
* clear(): Removes all the elements from the list.
* in and not in: Check if an element is present or not present in the list.

```python
languages = ['C#', 'Java', 'C++']
languages
```

```python
languages.append('Python')
languages
```

```python
languages.extend(['Go', 'R'])
languages
```

```python
languages.insert(0, 'Cobol')
languages
```

```python
languages.remove('R')
languages
```

```python
languages.pop(0)
```

```python
languages.reverse()
languages
```

```python
'Python' in languages
```

```python
'SQL' not in languages
```

```python
languages.count('Python')
```

```python
languages.index('Python')
```

```python
my_languages = languages.copy()
my_languages
```

```python
languages.clear()
languages
```

# Dictionaries

Dictionaries are a built-in Python data structure for mapping keys to values.

```python
numbers = {'one':1, 'two':2, 'three':3}
numbers, type(numbers)
```

## Indexing

```python
numbers['one']
```

```python
'two' in numbers
```

```python
'ten' in numbers
```

## Changing dictionaries

```python
numbers['one'] = 'one'
numbers
```

```python
numbers['eleven'] = 11
numbers
```

## Dictionary Methods

```python
numbers.keys()
```

```python
numbers.values()
```

```python
numbers.items()
```

```python
numbers.get('two')
```

```python
numbers.update({'three': 'three'})
numbers
```

```python
numbers.update({'twenty': 20})
numbers
```

```python
numbers.pop('one')
```

```python
numbers.clear()
numbers
```

## Iterating Dictionaries

```python
numbers = {'one':1, 'two':2, 'three':3}
numbers
```

```python
for n in numbers:
    print(n)
```

```python
for n in numbers:
    print(numbers[n])
```

```python
for k, v in numbers.items():
    print(k, v)
```

# Data Structure Comprehensions

## List Comprehensions

```python
squares = []
for n in range(10):
    squares.append(n**2)
    
squares
```

```python
squares = [n**2 for n in range(10)]
squares
```

```python
squares = [n**2 for n in range(10) if n%2 == 0]
squares
```

## Dictionary Comprehensions

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
planet_to_initial = {planet: planet[0] for planet in planets}
planet_to_initial
```
